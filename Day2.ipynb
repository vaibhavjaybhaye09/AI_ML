```python
cae1_cv2 = cv2.imread(img_path)
cae1_cv2
```

```python
# convert the color
car1_cv2_BGR_RGB = cv2.cvtColor(cae1_cv2, cv2.COLOR_BGR2RGB)
plt.imshow(car1_cv2_BGR_RGB)
```

```python
# Convert the BGR image (loaded by OpenCV) to HSV color space
hsv_car = cv2.cvtColor(cae1_cv2, cv2.COLOR_BGR2HSV)
hsv_car

# print("Image converted to HSV color space.")
```

```python
# Define the lower and upper bounds for the red color in HSV
# Red color wraps around the 0-180 hue range in OpenCV (0-360 degrees usually)
# So, we usually define two ranges for red: a lower red and an upper red.

# Lower red range
lower_red1 = np.array([0, 120, 70])
upper_red1 = np.array([10, 255, 255])

# Upper red range
lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])

# Create masks for both red ranges
mask1 = cv2.inRange(hsv_car, lower_red1, upper_red1)
mask2 = cv2.inRange(hsv_car, lower_red2, upper_red2)

# Combine the masks to get the full red mask
red_mask = mask1 + mask2

plt.imshow(red_mask, cmap='gray')
plt.title('Red Mask of the Car')
plt.show()
```

```python
def visualize_RGB_channels(imgArray = None, figsize =(10,7)):
  #spilit the RGB components
  B,G,R = cv2.split(imgArray)

  #create zero matrix of shape of image
  Z = np.zeros(B.shape, dtype = B.dtype) # can use any channle

  # init subplots
  fig, ax = plt.subplots(2,4, figsize = figsize)

  # ploting the actual image and RGB images
  [axi.set_axis_off() for axi in ax.ravel()]

  ax[0,0].set_title("Oringinal Image")
  ax[0,0].imshow(cv2.merge((R,G,B)))

  ax[0,1].set_title("Red Channel image")
  ax[0,1].imshow(cv2.merge((R,Z,Z)))

  ax[0,2].set_title("Green Channel image")
  ax[0,2].imshow(cv2.merge((Z,G,Z)))

  ax[0,3].set_title("Blue Channel image")
  ax[0,3].imshow(cv2.merge((Z,Z,B)))
```

```python
visualize_RGB_channels(imgArray = car1_cv2)
```
