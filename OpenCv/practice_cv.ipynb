```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("images/sky.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.axis()
plt.show()


```

```python
cv2.imwrite("output.jpg",img)


h, w, c = img.shape
print(f"Height : {h}, Width: {w}, color channle : {c}")
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray)
plt.axis()
plt.show()
```

```python
resized_image = cv2.resize(img, None, fx=0.5, fy=0.5 )
plt.imshow(resized_image)
plt.axis("off")
plt.show()
resized_image.shape
```

```python
blur = cv2.GaussianBlur(resized_image,(9,9),0)
plt.imshow(blur)
plt.axis("off")
plt.show()
resized_image.shape
```

```python
crop = img[100:500, 50:150]
plt.imshow(crop)
plt.axis("off")
plt.show()
```

```python
# Rotate Image  using // to find the image centre point 
# Get image dimensions
(h, w) = img.shape[:2]
center = (w // 2, h // 2)

# Get rotation matrix
M = cv2.getRotationMatrix2D(center, 90, 1.0)

# Apply affine transformation
rotated = cv2.warpAffine(img, M, (w, h))

# Display
plt.imshow(rotated)
plt.axis("off")
plt.show()                         
```
